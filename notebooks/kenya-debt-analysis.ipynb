```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/cleaned/kenya_debt.csv")

print(df.shape)
print(df.head(10))
```

```python
df.describe().round(2)
```

```python
print(df.isnull().sum())
print()
```

```python
threshold_debt = 40
threshold_service = 15

years_above_debt = df[df["external_debt_gni"] > threshold_debt]["year"].tolist()
years_above_service = df[df["debt_service_exports"] > threshold_service]["year"].tolist()

print("Years external debt exceeded 40% threshold:")
print(years_above_debt)
print()
print("Years debt service exceeded 15% threshold:")
print(years_above_service)
print()
```

```python
crisis1_debt = [y for y in years_above_debt if y <= 2010]
crisis2_debt = [y for y in years_above_debt if y > 2010]

crisis1_service = [y for y in years_above_service if y <= 2010]
crisis2_service = [y for y in years_above_service if y > 2010]

print("First crisis period (external debt):", crisis1_debt[0], "to", crisis1_debt[-1])
print("Second crisis period (external debt):", crisis2_debt)
print()
print("First crisis period (debt service):", crisis1_service[0], "to", crisis1_service[-1])
print("Second crisis period (debt service):", crisis2_service[0], "to", crisis2_service[-1])
```

```python
warning_years = df[(df["year"] >= 2014) & (df["year"] <= 2022)]

print(warning_years[["year", "external_debt_gni", "debt_service_exports", "gdp_growth"]].to_string())
```

```python
change_2017_2019 = df[df["year"] == 2019]["debt_service_exports"].values[0] - df[df["year"] == 2017]["debt_service_exports"].values[0]

print(f"Debt service to exports in 2017: {df[df['year'] == 2017]['debt_service_exports'].values[0]:.1f}%")
print(f"Debt service to exports in 2019: {df[df['year'] == 2019]['debt_service_exports'].values[0]:.1f}%")
print(f"Change in two years: +{change_2017_2019:.1f} percentage points")
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(df["year"], df["external_debt_gni"], color="black", linewidth=2)

ax.axhline(y=40, color="red", linestyle="--", linewidth=1.2, label="IMF threshold (40%)")

ax.fill_between(df["year"], df["external_debt_gni"], 40,
                where=(df["external_debt_gni"] > 40),
                alpha=0.2, color="red", label="Above threshold")

ax.set_title("Kenya External Debt to GNI (1980-2024)")
ax.set_xlabel("Year")
ax.set_ylabel("External Debt (% of GNI)")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(df["year"], df["external_debt_gni"], color="black", linewidth=2)

ax.axhline(y=40, color="red", linestyle="--", linewidth=1.2, label="IMF threshold (40%)")

ax.fill_between(df["year"], df["external_debt_gni"], 40,
                where=(df["external_debt_gni"] > 40),
                alpha=0.2, color="red", label="Above threshold")

ax.annotate("Structural adjustment peak",
            xy=(1993, 131.9),
            xytext=(1996, 118),
            arrowprops=dict(arrowstyle="->", color="black"),
            fontsize=8)

ax.annotate("HIPC debt relief",
            xy=(2000, 87.8),
            xytext=(2003, 100),
            arrowprops=dict(arrowstyle="->", color="black"),
            fontsize=8)

ax.annotate("First Eurobond",
            xy=(2014, 27.9),
            xytext=(2011, 45),
            arrowprops=dict(arrowstyle="->", color="black"),
            fontsize=8)

ax.annotate("COVID borrowing",
            xy=(2020, 38.0),
            xytext=(2017, 55),
            arrowprops=dict(arrowstyle="->", color="black"),
            fontsize=8)

ax.set_title("Kenya External Debt to GNI (1980-2024)")
ax.set_xlabel("Year")
ax.set_ylabel("External Debt (% of GNI)")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(df["year"], df["debt_service_exports"], color="black", linewidth=2)

ax.axhline(y=15, color="red", linestyle="--", linewidth=1.2, label="IMF threshold (15%)")

ax.fill_between(df["year"], df["debt_service_exports"], 15,
                where=(df["debt_service_exports"] > 15),
                alpha=0.2, color="red", label="Above threshold")

ax.annotate("Warning signal",
            xy=(2018, 23.7),
            xytext=(2015, 30),
            arrowprops=dict(arrowstyle="->", color="black"),
            fontsize=8)

ax.annotate("Peak 38.4%",
            xy=(2019, 38.4),
            xytext=(2016, 38),
            arrowprops=dict(arrowstyle="->", color="black"),
            fontsize=8)

ax.set_title("Kenya Debt Service to Exports (1980-2024)")
ax.set_xlabel("Year")
ax.set_ylabel("Debt Service (% of Exports)")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(df["year"], df["gdp_growth"], color="black", linewidth=2)

ax.axhline(y=0, color="red", linestyle="--", linewidth=1, label="Zero growth line")

ax.fill_between(df["year"], df["gdp_growth"], 0,
                where=(df["gdp_growth"] < 0),
                alpha=0.2, color="red", label="Negative growth")

ax.annotate("COVID recession",
            xy=(2020, -0.27),
            xytext=(2017, -1.5),
            arrowprops=dict(arrowstyle="->", color="black"),
            fontsize=8)

ax.set_title("Kenya GDP Growth (1980-2024)")
ax.set_xlabel("Year")
ax.set_ylabel("GDP Growth (%)")
ax.legend()
plt.tight_layout()
plt.show()
```
